```python
import os; print(os.popen('pip install hapiclient --upgrade').read())

from hapiclient import hapi

server     = 'https://cdaweb.gsfc.nasa.gov/hapi'
dataset    = 'OMNI2_H0_MRG1HR'
# Notes:
# 1. Use parameters='' to request all parameters from OMNI2_H0_MRG1HR.
# 2. Multiple parameters can be requested using a comma-separated
#    list, e.g., parameters='Rot1800,IMF1800'
parameters = 'BZ_GSM1800,V1800,DST1800'
start      = '1963-01-01T00:00:00Z' # min 1963-01-01T00:00:00Z
stop       = '2026-01-13T16:00:00Z' # max 2026-01-13T16:00:00Z

data, meta = hapi(server, dataset, parameters, start, stop)
```

```python
import pandas as pd

df = pd.DataFrame(data)
```

```python
df['Time'] = df['Time'].str.decode('utf-8')
df['time1'] = pd.to_datetime(df['Time'])
df['bz_gsm'] = df['BZ_GSM1800']
df['v'] = df['V1800']
df['dst'] = df['DST1800']
```

```python
df = df.drop(columns = ['Time', 'BZ_GSM1800', 'V1800', 'DST1800'])
```
