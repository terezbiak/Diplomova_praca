# Graf pre DST+6

Notebook porovnava nove kontinualne a event experimenty pre horizont DST+6 v zadanom casovom okne.

```python
from pathlib import Path

from viz_nove_experimenty import (
    OUTPUT_ROOT,
    plot_continuous_vs_event_window,
    save_continuous_summary_csv,
    save_event_summary_csv,
    summarize_continuous_predictions,
    summarize_event_predictions,
)
```

```python
continuous_variant = "DST+BZ_GSM+V"
event_feature_key = "dst_bz_v"
start_date = "2024-12-31"
end_date = "2025-01-03"
horizon_hours = 6

output_dir = OUTPUT_ROOT / "graf_DST+6"
output_dir.mkdir(parents=True, exist_ok=True)
output_dir
```

```python
continuous_summary = summarize_continuous_predictions(continuous_variant)
event_summary = summarize_event_predictions()

continuous_csv = None
event_csv = None
if not continuous_summary.empty:
    continuous_csv = save_continuous_summary_csv(continuous_summary, output_dir)
if not event_summary.empty:
    event_csv = save_event_summary_csv(event_summary, output_dir)

plot_path = output_dir / f"continuous_vs_event_DST+{horizon_hours}.png"
plot_continuous_vs_event_window(
    continuous_variant=continuous_variant,
    event_feature_key=event_feature_key,
    horizon_hours=horizon_hours,
    start_date=start_date,
    end_date=end_date,
    output_path=plot_path,
)

print(f"Output directory: {output_dir}")
print(f"Plot: {plot_path}")
print(f"Continuous summary CSV: {continuous_csv}")
print(f"Event summary CSV: {event_csv}")

if not continuous_summary.empty:
    display(
        continuous_summary[continuous_summary["horizon_hours"] == horizon_hours]
        .sort_values(["lookback_hours"])
        .reset_index(drop=True)
    )
if not event_summary.empty:
    display(
        event_summary[event_summary["horizon_hours"] == horizon_hours]
        .sort_values(["feature_key", "lookback_hours"])
        .reset_index(drop=True)
    )
```
